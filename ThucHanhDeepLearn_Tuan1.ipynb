```python
@tf.function
def perimeter(a, b):
    return (a + b) * 2

print("B15:", perimeter(5.0, 3.0))
@tf.function
def surface_area(a, b, c):
    Sxq = 2 * c * (a + b)
    Stp = 2 * (a*b + b*c + a*c)
    return Sxq, Stp

print("B16:", surface_area(2.0, 3.0, 4.0))
```

#bai19

```python
# AM = MB, AN = NC => MN // BC và MN = BC/2

@tf.function
def midline(AM, MB, AN, NC, BC):
    condition = tf.logical_and(tf.equal(AM, MB), tf.equal(AN, NC))
    MN = BC / 2
    return condition, MN

cond, MN = midline(2.0,2.0,3.0,3.0,10.0)
print("B19:")
print("Điều kiện:", cond.numpy())
print("MN =", MN.numpy())
```
